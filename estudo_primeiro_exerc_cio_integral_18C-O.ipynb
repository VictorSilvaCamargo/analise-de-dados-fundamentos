# Questão 1

Mesmo que você ainda não entenda o que é uma integral deve ser capaz de resolver o seguinte problema:
Você é um engenheiro de sistemas que trabalha na otimização de um algoritmo de processamento de sinal digital. O sinal é modelado pela função $f(t)= t^2 \times cos(t)$, onde $t$ representa o tempo em segundos. Você está interessado na quantidade total de sinal processado entre os tempos $t=1s$ e $t=4s$, que pode ser calculada como a integral definida da função $f(x)$ entre esses limites.

Sua Tarefa é:

1. Calcular a integral definida da função $f(t)= t^2 \times cos(t)$ entre $t=1s$ e $t=4s$.
2. Plote a função $f(t)$ para $t$ entre $0s$ e $5s$, usando o matplotlib e o sympy no Google Colab.
3. Marque os limites da integral definida ($t=1$ e $t=4$) com linhas pontilhadas verticais em azul.
4. Preencha a área sob a curva entre $t=1$ e $t=4$, que representa a quantidade total de sinal processado durante esse período de tempo.
5. Imprima na legenda a função, a integral definida, e o seu valor com latex.








```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import symbols, cos, integrate, lambdify

# 1. Calcula a integral definida da função f(t) = t^2 * cos(t) entre t=1 e t=4

# Define a variável e a função
t = symbols('t')
f_sym = t**2 * cos(t)

# Calcula a integral definida
integral = integrate(f_sym, (t, 1, 4))
integral_value = float(integral)

print(f'Valor da integral definida: {integral_value}')

# 2. Plota a função f(t) para t entre 0 e 5 segundos

# Converte f para uma função numérica para plotagem
f = lambdify(t, f_sym, 'numpy')

t_values = np.linspace(0, 5, 400)
f_values = f(t_values)

plt.figure(figsize=(10, 6))
plt.plot(t_values, f_values, label='$f(t) = t^2 \cdot cos(t)$')

# 3. Marca os limites da integral definida (t=1 e t=4) com linhas pontilhadas verticais em cinza claro
plt.axvline(x=1, color='blue', linestyle='--')
plt.axvline(x=4, color='blue', linestyle='--')

# E anota os pontos t=1 e t=4
f1 = f(1)
f4 = f(4)
plt.scatter([1, 4], [f1, f4], color='red')  # marca os pontos com círculos vermelhos
plt.annotate(f'({1}, {round(f1, 2)})', (1, f1), textcoords="offset points", xytext=(26,10), ha='center')
plt.annotate(f'({4}, {round(f4, 2)})', (4, f4), textcoords="offset points", xytext=(30,-14), ha='center')

# 4. Preenche a área sob a curva entre t=1 e t=4
idx = np.where((t_values >= 1) & (t_values <= 4)) # encontra os índices para t entre 1 e 4
plt.fill_between(t_values[idx], f_values[idx], color='lightgray', alpha=0.3)

# finalmente mostra o gráfico
integral_label = '$\int_1^4 t^2 \cdot cos(t) \, dt = ' + str(round(integral_value, 2)) + '$'
plt.legend([integral_label])
plt.xlabel('Tempo (t)')
plt.ylabel('f(t)')
plt.title('Minha Primeira Integral')
plt.grid(True)
plt.show()
```
