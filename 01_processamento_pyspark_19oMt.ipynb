```python
import pyspark
from pyspark.sql import SparkSession
```

```python
# criação de uma sessão spark
spark = SparkSession.builder.appName("Word Count").master('local[*]').getOrCreate()

# criação do SparkContext (conexão entre o Driver e Workers)
sc = spark.sparkContext
```

```python
rdd = sc.textFile("Airports.csv")
```

```python
rdd_br = rdd.filter(lambda x: x.split(",")[4] == "BRAZIL")
rdd_br = rdd_br.filter(lambda x: float(x.split(",")[13]) > 25)

rdd_lower = rdd_br.map(lambda x: x.lower())

rdd_decre = rdd_lower.sortBy(lambda x: x.split(",")[13], ascending=False)

filtro = rdd_decre.collect()

# Exibir os resultados
for i in filtro:
    print(i)
```

```python
rdd_decre.coalesce(1).saveAsTextFile("airport_Br.csv")
```

```python
rdd = sc.textFile("airport_Br.csv")

altitudes = rdd.map(lambda x: float(x.split(",")[13]))

total_altitude = altitudes.sum()
count = altitudes.count()
average_altitude = total_altitude / count

print(f"{average_altitude:.2f}")
```

```python
altitude = rdd.map(lambda x: float(x.split(",")[13]))

menor_altitude = altitude.min()
print(menor_altitude)

maior_altitude = altitude.max()
print(maior_altitude)
```
