```python
curitiba = tabela.loc[(tabela["Cidade e UF"] == "Curitiba - PR") & (tabela["Ano"] == 2023)]
display(curitiba)
```

```python
import numpy as np
from tabulate import tabulate

curitiba['Valor'] = curitiba['Valor'].astype(str).str.replace(',', '').astype(float)

num_classes = 5

valor_min = curitiba['Valor'].min()
valor_max = curitiba['Valor'].max()

intervalos = np.linspace(valor_min, valor_max + 0.001, num_classes + 1)

frequencias = [(f'{intervalos[i]:.6f} - {intervalos[i+1]:.6f}', ((curitiba['Valor'] >= intervalos[i]) & (curitiba['Valor'] < intervalos[i+1])).sum()) for i in range(num_classes)]

frequencias = [(intervalo, int(frequencia)) for intervalo, frequencia in frequencias]

pontos_medios = [(intervalos[i] + intervalos[i+1]) / 2 for i in range(num_classes)]

frequencias = [(f'{intervalos[i]:.6f} - {intervalos[i+1]:.6f}', pontos_medios[i], int(frequencia)) for i, (intervalo, frequencia) in enumerate(frequencias)]

frequencia_acumulada = 0
frequencias_acumuladas = []
for intervalo, ponto_medio, frequencia in frequencias:
    frequencia_acumulada += frequencia
    frequencias_acumuladas.append((intervalo, ponto_medio, frequencia, frequencia_acumulada))

print(tabulate(frequencias_acumuladas, headers=['frequencias', 'Ponto Médio', 'Intervalo de Classe', 'Frequência Acumulada'], tablefmt='grid'))

```
