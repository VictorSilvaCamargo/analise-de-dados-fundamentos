# Resolução questão 1

```python
frase = "O processamento de linguagem natural permite a análise de grandes volumes de texto"

def tokenizar(frase):
  stopwords = ["o","de","a","em"]
  frase_token = frase.split()
  frase_filtrada = []

  for palavra in frase_token:
    if palavra not in stopwords:
      frase_filtrada.append(palavra)


  return frase_filtrada

tokens = tokenizar(frase)
print(tokens)
```

# Resolução questão 6

```python
# Renan Antonio Hammerschmidt Krefta
# Victor Silva Camargo
# Vinícius Silva Camargo
import nltk
nltk.download('punkt_tab')
nltk.download('rslp')
import math
from nltk.tokenize import word_tokenize
d1 = "O jogador marcou dois gols na partida"
d2 = "A atleta fez uma grande atuação no jogo"
d3 = "O clube venceu o campeonato nacional"
corpus = [d1, d2, d3]
# Bag of words
# adquirindo os termos
termos = []
for documento in corpus:
 for termo in documento.split():
   if termo not in termos:
     termos.append(termo)
# criando uma matriz contendo vetores de frequencia para cada termo
matriz = []
for documento in corpus:
 vetor = []
 for termo in termos:
   vetor.append(documento.split(' ').count(termo))
 matriz.append(vetor)
# TF - IDF
# calculando TF
matriz_tf = []
for documento in corpus:
 vetor = []
 for termo in termos:
   vetor.append(documento.split(' ').count(termo) / len(documento.split(' ')))
 matriz_tf.append(vetor)
# calculando idf com log na base 10
vetor_idf = []
tamanho_corpus = len(corpus)
for termo in termos:
 # numero de documentos com o termo
 documentos_com_o_termo = 0
 for documento in corpus:
   if termo in documento.split(' '):
     documentos_com_o_termo += 1
 vetor_idf.append(math.log10(tamanho_corpus / documentos_com_o_termo))
```

```python
matriz_tf_idf = []
for i in range(len(matriz_tf)):
 vetor = []
 vetor2 = []
 for j in range(len(matriz_tf[i])):
   vetor.append(matriz_tf[i][j] * vetor_idf[j])
 matriz_tf_idf.append(vetor)



for termo in range (len(termos)):
  variavel = termos[termo]
  nltk.RSLPStemmer().stem(variavel)
  vetor2.append(variavel)

print(vetor2)
```
