```python
import random
from tabulate import tabulate

letras1 = ['q', 'w', 'x', 'z']
letras2 = ['a', 'i', 'u']
letras3 = ['c', 'f', 'p']
letras4 = ['e', 'o']

todas_palavras = []

for letra1 in letras1:
    for letra2 in letras2:
        for letra3 in letras3:
            for letra4 in letras4:
                palavra = letra1 + letra2 + letra3 + letra4
                todas_palavras.append([palavra])


num_sorteios = [72, 216, 720, 2160, 7200, 72000]


sorteios = {}
for num in num_sorteios:
    sorteios[num] = [random.choice(todas_palavras)[0] for _ in range(num)]


for num in num_sorteios:
    coluna = f"{num} sorteios"
    for palavra in todas_palavras:
        contador = sorteios[num].count(palavra[0])
        palavra.append(contador)


esperado_percentual = round((1 / 72) * 100, 2)
for palavra in todas_palavras:
    palavra.append(esperado_percentual)


print(tabulate(todas_palavras, headers=["Palavra", *map(lambda x: f"{x} sorteios", num_sorteios), "Esperado (%)"]))

print("O número de vezes que cada palavra sorteada foi igual em cada conjunto de sorteios?\n")
print("Resposta = Não\n")
print("Em todas as colunas os números são iguais percentualmente? Por quê?\n")
print("Resposta = Sim, pois por mais que a quantidade de geração aleatória de palavras multiplique, o numero de palavras possiveis e o mesmo, no caso 72\n")
print("Há alguma tendência à medida que se realizam mais sorteios? Em caso positivo, qual?\n")
print("Resposta = Sim, aumenta a quantidade de vezes que uma palavra e sorteada\n")
print("O que se pode concluir deste experimento?\n")
print("Resposta = A quantidade de vezes que cada deveria aparecer e a mesma, mesmo aumentando o numero de sorteios, devido a ser somente 72 palavras a serem sorteadas,")
print("o que muda e a quantidade de vezes que ela aparece em função do numero de sorteios")
```
