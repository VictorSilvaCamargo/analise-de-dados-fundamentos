```
from pyspark import SparkContext, SparkConf
from tabulate import tabulate

# Substitua "caminho/do/arquivo/cadastro_de_empregadores.csv" pelo caminho real do seu arquivo
caminho_arquivo = "cadastro_de_empregadores.csv"

# Lê o arquivo CSV como um RDD
rdd = sc.textFile(caminho_arquivo)

# Use a ação "take" para obter as primeiras 5 linhas
primeiras_linhas = rdd.take(5)

# Exibe as primeiras 5 linhas em forma de tabela
print("Ocorrência de trabalhos análogos a escravidão no Brasil")
print("Fonte: Ministério do Trabalho e do Emprego")
print("Inicio da coleta de dados: Sem Informação")
print("Fim da coleta de dados: Não acabou a coleta de dados")
print("Peculiaridade: A tabela é atualizada a cada semestre, próxima atualização está prevista para dia 02/04/2024")
print("Existem 10 variáveis, sendo elas:\n")
print("Ano da ação fiscal, Tipo: Qualitativa Nominal.\n")
print("UF, Tipo: Qualitativa Nominal\n")
print("Empregador, Tipo: Qualitativa Nominal \n")
print("CNPJ/CPF, Tipo: Qualitativa Nominal \n")
print("Estabelecimento, Tipo: Qualitativa Nominal \n")
print("Trabalhadores envolvidos, Tipo: Qualitativa Nominal \n")
print("CNAE, Tipo: Qualitativa Nominal \n")
print("Decisão administrativa de procedência, Tipo: Qualitativa Nominal \n")
print("Inclusão no Cadastro de Empregadores, Tipo: Qualitativa Nominal \n")
print("A 'lista suja' ja foi tema de materias do G1 e do própio governo federal, postados em suas paginas de notícia, o intuito das reportagens e expor os empregadores que submetem seus\n")
print("empregados a trabalhos análogos a escavidão, determinar em quais estados estão a maioría dos casos, etc.")



# Criar uma lista de listas para os dados
dados_formatados = [linha.split(",") for linha in primeiras_linhas]

# Imprimir tabela usando tabulate
print(tabulate(dados_formatados, tablefmt="grid"))

# Extrai a terceira coluna (índice 2) e faz a contagem de cada estado
contagem_estados = rdd.map(lambda linha: linha.split(",")[2]).filter(lambda estado: estado != "UF")
contagem_estados = contagem_estados.map(lambda estado: (estado, 1))
contagem_estados = contagem_estados.reduceByKey(lambda x, y: x + y)
contagem_estados = contagem_estados.sortBy(lambda x: x[1])

# Coleta os resultados
resultados = contagem_estados.collect()

# Imprime os resultados em forma de tabela
print(tabulate(resultados, headers=["Estado", "Contagem"], tablefmt="grid"))
```
